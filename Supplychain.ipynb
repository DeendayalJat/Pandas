```python

import numpy as np
import pandas as pd
```

```python
df=pd.read_csv('/content/supply_chain - supply_chain.csv')
df.head(2)
```

```python
df.isnull().sum()
```

```python
df['Product type'].unique()
```

```python
import plotly.express as px
import plotly.io as pio
import plotly.graph_objects as go
pio.templates.default ='plotly_white'
```

```python
fig=px.scatter(df,
               x='Price',
               y='Revenue generated',
               color='Product type',
               hover_data=['Number of products sold'],
               trendline='ols')
fig.show()
```

```python
sales_data=df.groupby('Product type')['Number of products sold'].sum().reset_index()
sales_data
```

```python
pie_chart=px.pie(sales_data,
                 names='Product type',
                 values='Number of products sold',
                 title='Sales by Product Type',
                 hover_data=['Number of products sold'],
                 hole=0.5,
                 color_discrete_sequence=px.colors.qualitative.Pastel
                 )
pie_chart.update_traces(textposition='inside',textinfo='percent+label')
pie_chart.show()
```

```python
total_revenue=df.groupby('Shipping carriers')['Revenue generated'].sum().reset_index()
total_revenue
```

```python
fig=go.Figure()
fig.add_trace(go.Bar(
    x=total_revenue['Shipping carriers'],
    y=total_revenue['Revenue generated'],
))
fig.update_layout(
    title='Total Revenue by Shipping Carriers',
    xaxis_title='Shipping Carriers',
    yaxis_title='Total Revenue'
)
fig.show()
```

```python
deffect_target=df.groupby('Product type')['Defect rates'].sum().reset_index()
deffect_target
```

```python
pie_chart1=px.pie(deffect_target,
                 names='Product type',
                 values='Defect rates',
                 title='Defect Rates by Product Type',
                 hover_data=['Defect rates'],
                 hole=0.5,
                 #color_discrete_sequence=px.colors.qualitative.Pastel
                 )
pie_chart1.update_traces(textposition='inside',textinfo='percent+label')
pie_chart1.show()
```

```python
avg_shipping_time=df.groupby('Product type')['Shipping times'].mean().reset_index()
avg_shipping_time
```

```python
fig=px.pie(avg_shipping_time,
           names='Product type',
           values='Shipping times',
           title='Average Shipping Time by Product Type',
           hover_data=['Shipping times'],
           hole=0.3,
           color_discrete_sequence=px.colors.qualitative.Pastel
           )
fig.update_traces(textposition='inside',textinfo='percent+label')
fig.show()
```

```python
avg_lead_time=df.groupby('Product type')['Lead times'].mean().reset_index()
avg_lead_time
```

```python
avg_manufacturing_costs=df.groupby('Product type')['Manufacturing costs'].mean().reset_index()
avg_manufacturing_costs
```

```python
result=pd.merge(avg_manufacturing_costs,avg_lead_time,on='Product type')
#result.rename(columns={'Lead times':'Average Lead Time','Manfacturing costs':'Average Manufacturing Costs'},inplace=True)
result.rename(columns={'Lead times':'Average Lead Time','Manufacturing costs':'Average Manufacturing Costs'},inplace=True)
result
```
